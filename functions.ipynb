# functions

#### functions provide a way to reuse code. Functions can be called at any part of the code

## def and return
#### in python, we use the def keyword to create a function followed by the name of the function.
#### we use return statements to give the output of the function.

## example 1
#### a function that retunrs a message "i love python".
#### note that python allows us to return value without specifying the data type (dynamic typing).


```python
def message():
    return 'i love python'

# calling the function.
print(message()) # outputs a stirng i love python
```

## function parameters and arguments
#### funcitons take parameters which represent values passed to the function.
#### these values passed to the function when calling it are called arfuments.

## example 2
#### a function which takes two parameters and outputs the sum of two integers.

```python
def summation(a, b):
    return a + b

print(summation(5, 6)) # output 11
```

## default values
#### sometimes we can use default values for parameters which are used when an argument is not passed to the function.

```python
def multiplication(a, b=2):
    return a * b

print(multiplication(4)) # output 8
print(multiplication(4, 3)) # output 12
```

## variable scope
#### when variables are initialized in a function, changes made to them within that function remain in the function

## example 3
#### a varible a which is passed to a function which increments its value by 2

```python
a = 2

def scopeDemo(a):
    a += 2 # increment a by 2
    return a

print(a) #output 2
print(scopeDemo(a)) # output 4
```

## recursive functions
#### functions that make calls to themselves
#### they help to reduce number of steps of loops in functions

## example 4 (factorial)
#### a function takes an integer as an argument and returns the factorial
#### a factorial a! = a * (a-1) * (a-2) * ..... * 1

```python
def factorial(a):
    if a == 1: # base case to stop multiplication when a becomes 1
        return a
    else:
        return a * factorial(a - 1) # the recursive step which keeps calling the function if a is not equal to 1
    
print(factorial(3)) # output 6
```
